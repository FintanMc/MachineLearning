### Target
The target of this portfolio is `depression_label`

```python
depressed = lambda x: Teen_Mental_Health_Dataset.loc[(Teen_Mental_Health_Dataset.depression_label == 0) ^ x].count()["depression_label"]

display(pd.DataFrame({
  "depression_label = true": depressed(True),
  "depression_label = false": depressed(False)
}, index=["count"]).transpose())
```

As you can see, over 97% of the rows have a false depression label, however this does not render the dataset unusable. I now know that i need to avoid `accuracy` as a metric because i would expect around 97% accuracy for even a baseline model, instead i should look at `recall`, `precision` or combine the 2 using their harmonic mean `f1`. Since the number of positive cases is quite low, stratifying the dataset for things like cross validation may be difficult.

Lastly, I will give the binary representation of `depression_label` a categorical label so that it is easier to interpret, and save the dataset to the notebook store to access in other notebooks

```python
Teen_Mental_Health_Dataset["depression_label"] = Teen_Mental_Health_Dataset["depression_label"].apply(
    lambda x: "depressed" if x == 1 else "not depressed"
)

%store Teen_Mental_Health_Dataset
```
