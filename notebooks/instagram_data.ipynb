```python
import pandas as pd
df = pd.read_csv('../data/instagram_data.csv')

df
```

```python
from sklearn.tree import DecisionTreeRegressor

column_to_test = 'virality_score'
model = DecisionTreeRegressor()
number_of_models = 10

data = df.copy()

data = data.filter(items=[
'reach',
'impressions',
'non_follower_reach_ratio',
'explore_page_boost',
column_to_test,
])

print(df.columns)
data
```

```python
from sklearn.model_selection import train_test_split

X = data.drop(columns=[column_to_test])
y = data[column_to_test]

print("Features (X):")
print(X)
print("Target (y):")
print(y)
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

model.fit(X_train, y_train)
y_pred = model.predict(X_test)
y_train_pred = model.predict(X_train)

training_result = pd.DataFrame({'Actual': y_train, 'Predicted': y_train_pred})
test_result = pd.DataFrame({'Actual': y_test, 'Predicted': y_pred})

training_accuracy = model.score(X_train, y_train)
print(f"Training Accuracy: {training_accuracy}")
print(training_result)

test_accuracy = model.score(X_test, y_test)
print(f"Test Accuracy: {test_accuracy}")
print(test_result)
```
