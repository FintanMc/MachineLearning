```python
from sklearn.tree import DecisionTreeClassifier
from create_model import create_model
from grid_cross_validation import grid_cross_validation

decisionTree = create_model(DecisionTreeClassifier(random_state=environment.Random_State))

cv = grid_cross_validation(decisionTree, X_train, y_train, param_grid={
    'decisiontreeclassifier__max_depth': [None, 5, 10, 15],
    'decisiontreeclassifier__min_samples_split': [2, 5, 10],
    'decisiontreeclassifier__min_samples_leaf': [1, 2, 4],
})

display(pd.DataFrame([cv.best_params_], index=["Best Parameters"]).T)
#display(pd.DataFrame(cv.cv_results_))
best_estimator = cv.best_estimator_.fit(X_train, y_train)
```

```python
from run_test import run_test

confusion_matrix, classification_report = run_test(best_estimator, splits)

display(confusion_matrix)
print(classification_report)
```
