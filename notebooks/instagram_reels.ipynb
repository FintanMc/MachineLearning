```python
import pandas as pd

df = pd.read_csv('../data/instagram_reels_clean.csv')

column_to_test = 'virality_score'

df = df.drop(columns=[
'post_id',
'shortcode',
'views',
'likes',
'comments',
'creator_handle',
'timestamp_utc',
'caption',
'engagement_rate_pct',
'virality_tier'
])

df
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsRegressor
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler

model = make_pipeline(
    ColumnTransformer(
        transformers=[
            ('encoder', OneHotEncoder(handle_unknown="ignore", sparse_output=False), ['niche']),
        ]
    ),
    StandardScaler(),
    #LogisticRegression()
    KNeighborsRegressor(n_neighbors=5)
)
```

```python
X = df.drop(columns=[column_to_test])
y = df[column_to_test]
```

```python
from sklearn.model_selection import cross_val_score, train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

model.fit(X_train, y_train)

cross_val_scores = cross_val_score(model, X, y, cv=5)

test_score = model.score(X_test, y_test)

print(f"CV Accuracy: {cross_val_scores.mean()}")
print(f"CV Standard deviation: {cross_val_scores.std()}")

print(f"Test Accuracy: {test_score}")

predictions = pd.DataFrame({
  "actual": y_test,
  "predicted": model.predict(X_test)
})
predictions
```
