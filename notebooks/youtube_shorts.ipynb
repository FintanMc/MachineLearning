```python
import pandas as pd
df = pd.read_csv('../data/youtube_shorts_performance_dataset.csv')

df
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import LabelEncoder

column_to_test = 'engagement'
model = LinearRegression()

df["category"] = LabelEncoder().fit_transform(df["category"])
df["engagement"] = df.apply(lambda row: 100*(row["likes"] + row["comments"]*5 + row["shares"] * 10)/row["views"], axis=1)

df = df.drop(columns=[
"video_id",
"title",
"views",
"likes",
"comments",
"shares"
])

df
```

```python
from sklearn.model_selection import train_test_split

X = df.drop(columns=[column_to_test])
y = df[column_to_test]

print("Features (X):")
print(X)
print("Target (y):")
print(y)
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

model.fit(X_train, y_train)
y_pred = model.predict(X_test)

result = pd.DataFrame({'Actual': y_test, 'Predicted': y_pred})

accuracy = model.score(X_test, y_test)
print(f"Accuracy: {accuracy}")

print(result)
```
